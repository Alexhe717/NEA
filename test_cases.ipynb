```python
import board
from minimax import best_move

a = board.Board(5)
a.condition = 5 


a.change_state(0,0,1)
a.change_state(1,1,1)
a.change_state(2,2,1)
a.change_state(3,3,1)


a.change_state(0,4,2)
a.change_state(1,4,2)
a.change_state(4,0,2)

a.display()

mv = best_move(a, 1)
print("best_move=", mv)
```
